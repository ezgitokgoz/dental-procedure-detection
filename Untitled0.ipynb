## FILTRE

```python
#caries (class:0), filling (class:2), rootCanal (class:9) olan ve boneLoss (class: 13) olmayan
import os
import shutil

# Kaynak ve hedef klasörleri tanımlayın
source_label_folder = "/content/drive/MyDrive/dataset/labels/train/"  # Etiket dosyalarının bulunduğu klasör
source_image_folder = "/content/drive/MyDrive/dataset/images/train/"  # Görsellerin bulunduğu klasör
output_label_folder = "/content/drive/MyDrive/DemoColab/spesifikEtiketlerCalismasi/dataset/labels/train/"  # Filtrelenmiş etiketlerin kaydedileceği klasör
output_image_folder = "/content/drive/MyDrive/DemoColab/spesifikEtiketlerCalismasi/dataset/images/train/"  # Filtrelenmiş görsellerin kaydedileceği klasör


# Filtrelemek istediğiniz sınıflar
required_classes = {0, 2, 9}
excluded_class = 13

# Kaynak klasördeki tüm dosyaları işle
for file_name in os.listdir(source_label_folder):
    if file_name.endswith(".txt"):
        label_path = os.path.join(source_label_folder, file_name)
        image_name = os.path.splitext(file_name)[0] + ".jpg"  # Görsel adını varsayılan olarak ".jpg" kabul ediyoruz
        image_path = os.path.join(source_image_folder, image_name)

        # Dosyayı oku
        with open(label_path, "r") as file:
            lines = file.readlines()

        # Sınıfları kontrol et
        classes_in_file = set()
        contains_excluded_class = False

        for line in lines:
            parts = line.split()
            class_id = int(parts[0])  # İlk eleman sınıfı temsil eder

            if class_id == excluded_class:
                contains_excluded_class = True
                break

            classes_in_file.add(class_id)

        # Şartları kontrol et
        if contains_excluded_class:
            continue

        if required_classes.issubset(classes_in_file):
            # Yeni etiketi kaydet
            output_label_path = os.path.join(output_label_folder, file_name)
            with open(output_label_path, "w") as output_file:
                output_file.writelines(lines)

            # Görseli kaydet
            if os.path.exists(image_path):
                output_image_path = os.path.join(output_image_folder, image_name)
                shutil.copy(image_path, output_image_path)

print(f"Filtreleme tamamlandı. Etiketler {output_label_folder} klasörüne, görseller {output_image_folder} klasörüne kaydedildi.")

```

## YOLO FORMAT

```python
import os

def convert_gt_to_yolo(gt_file_path, yolo_file_path):
    with open(gt_file_path, 'r') as gt_file, open(yolo_file_path, 'w') as yolo_file:
        for line in gt_file:
            data = line.strip().split()
            class_id = data[0]
            coords = list(map(float, data[1:]))

            x_coords = coords[0::2]
            y_coords = coords[1::2]

            x_min, x_max = min(x_coords), max(x_coords)
            y_min, y_max = min(y_coords), max(y_coords)

            x_center = (x_min + x_max) / 2
            y_center = (y_min + y_max) / 2
            width = x_max - x_min
            height = y_max - y_min

            yolo_file.write(f"{class_id} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}\n")

# Örnek kullanım
input_folder = "/content/drive/MyDrive/DemoColab/donem2/donem2.1/dataset/test/ground_truths1"
output_folder = "/content/drive/MyDrive/DemoColab/donem2/donem2.1/dataset/test/ground_truths"

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

for gt_file in os.listdir(input_folder):
    if gt_file.endswith(".txt"):
        gt_path = os.path.join(input_folder, gt_file)
        yolo_path = os.path.join(output_folder, gt_file)
        convert_gt_to_yolo(gt_path, yolo_path)
        print(f"Converted {gt_file} to YOLO format.")
```

## BAŞARI METRİKLERİ

```python
import os
import numpy as np
import pandas as pd  # Pandas kütüphanesini içe aktar

# IoU hesaplama fonksiyonu
def calculate_iou(box1, box2):
    x1_min = box1[0] - box1[2] / 2
    y1_min = box1[1] - box1[3] / 2
    x1_max = box1[0] + box1[2] / 2
    y1_max = box1[1] + box1[3] / 2

    x2_min = box2[0] - box2[2] / 2
    y2_min = box2[1] - box2[3] / 2
    x2_max = box2[0] + box2[2] / 2
    y2_max = box2[1] + box2[3] / 2

    inter_x_min = max(x1_min, x2_min)
    inter_y_min = max(y1_min, y2_min)
    inter_x_max = min(x1_max, x2_max)
    inter_y_max = min(y1_max, y2_max)

    inter_area = max(0, inter_x_max - inter_x_min) * max(0, inter_y_max - inter_y_min)
    box1_area = (x1_max - x1_min) * (y1_max - y1_min)
    box2_area = (x2_max - x2_min) * (y2_max - y2_min)

    union_area = box1_area + box2_area - inter_area
    return inter_area / union_area if union_area > 0 else 0

# Dosyaları okuma ve sınıf bazında istatistik toplama
def evaluate_metrics(gt_folder, pred_folder, iou_threshold=0.5):
    classes = {}
    total_iou = []
    gt_files = sorted(os.listdir(gt_folder))
    pred_files = sorted(os.listdir(pred_folder))

    for gt_file, pred_file in zip(gt_files, pred_files):
        with open(os.path.join(gt_folder, gt_file), 'r') as gt_f, open(os.path.join(pred_folder, pred_file), 'r') as pred_f:
            gt_boxes = [line.strip().split() for line in gt_f.readlines()]
            pred_boxes = [line.strip().split() for line in pred_f.readlines()]

            gt_boxes = [(int(cls), list(map(float, box))) for cls, *box in gt_boxes]
            pred_boxes = [(int(cls), list(map(float, box))) for cls, *box in pred_boxes]

            used_preds = set()

            for gt_cls, gt_box in gt_boxes:
                if gt_cls not in classes:
                    classes[gt_cls] = {'tp': 0, 'fp': 0, 'fn': 0}
                matched = False
                for i, (pred_cls, pred_box) in enumerate(pred_boxes):
                    if i in used_preds or pred_cls != gt_cls:
                        continue
                    iou = calculate_iou(gt_box, pred_box)
                    if iou >= iou_threshold:
                        classes[gt_cls]['tp'] += 1
                        total_iou.append(iou)
                        used_preds.add(i)
                        matched = True
                        break
                if not matched:
                    classes[gt_cls]['fn'] += 1
            # False positives
            for i, (pred_cls, _) in enumerate(pred_boxes):
                if i not in used_preds:
                    if pred_cls not in classes:
                        classes[pred_cls] = {'tp': 0, 'fp': 0, 'fn': 0}
                    classes[pred_cls]['fp'] += 1

    # Precision, Recall, F1-score hesaplama
    results = {}
    for cls, stats in classes.items():
        tp, fp, fn = stats['tp'], stats['fp'], stats['fn']
        precision = tp / (tp + fp) if tp + fp > 0 else 0
        recall = tp / (tp + fn) if tp + fn > 0 else 0
        f1_score = 2 * precision * recall / (precision + recall) if precision + recall > 0 else 0
        results[cls] = {
            'precision': precision,
            'recall': recall,
            'f1_score': f1_score
        }

    mean_iou = np.mean(total_iou) if total_iou else 0
    return results, mean_iou

# Sonuçları tabloya dönüştürme
def display_results_as_table(results, mean_iou):
    # Sonuçları pandas DataFrame'e dönüştürme
    df = pd.DataFrame(results).T  # Transpose ederek sınıfları satırlara alıyoruz
    df['class'] = df.index  # 'class' sütununu sınıf ID'leriyle dolduruyoruz
    df = df[['class', 'precision', 'recall', 'f1_score']]  # Kolon sırasını düzenliyoruz

    # Genel IoU'yu tabloya ekliyoruz
    print("Sınıf Bazında Sonuçlar:")
    print(df)
    print(f"\nGenel IoU: {mean_iou:.4f}")

# Örnek kullanım
gt_folder = "/content/drive/MyDrive/DemoColab/donem2/dataset/test/ground_truths" #gerçek görüntü etiketleri
pred_folder = "/content/drive/MyDrive/DemoColab/donem2/test_results/testResults2/labels/" #test sonuçlarının etiketleri

results, mean_iou = evaluate_metrics(gt_folder, pred_folder)

# Sonuçları tablo olarak yazdır
display_results_as_table(results, mean_iou)
```
